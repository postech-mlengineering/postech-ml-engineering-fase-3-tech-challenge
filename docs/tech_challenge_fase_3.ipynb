# Tech Challenge - Fase 3: Análise e Predição de Atrasos de Voos

## 1. Introdução
O **Tech Challenge** é o projeto principal desta fase, integrando os conhecimentos de todas as disciplinas. Este trabalho foca na aplicação de técnicas de **Machine Learning Supervisionado e Não Supervisionado** utilizando um conjunto de dados real sobre o transporte aéreo nos EUA.

> **Atenção:** Esta é uma atividade obrigatória e representa **90% da nota final**.

---

## 2. O Problema
O transporte aéreo é vital para a infraestrutura global, mas atrasos impactam milhões de passageiros e geram custos elevados. O objetivo deste projeto é desenvolver análises e modelos preditivos/exploratórios para entender e antecipar esses atrasos.

*   **Base de Dados:** [Base de Dados - MLET Fase 3](https://drive.google.com/drive/folders/1aS7exW5N0qq1uIxvIBcAfc18OHojOMjj) (Substitua pelo link correto, se houver).

---

## 3. Objetivos do Projeto
Desenvolver um pipeline completo de ciência de dados, estimulando a criatividade e a exploração autônoma, abrangendo:
*   Exploração de dados (EDA).
*   Modelagem Supervisionada.
*   Modelagem Não Supervisionada.
*   Interpretação crítica de resultados.

---

## 4. Requisitos Obrigatórios (Checklist)

### 📊 Exploração dos Dados (EDA)
- [✅] Investigar os dados com estatísticas descritivas.
- [✅] Criar visualizações que tragam *insights* relevantes.
- [✅] Tratar valores ausentes de forma adequada.

### 🤖 Modelagem Supervisionada (Mínimo um modelo)
*Escolha entre Classificação (se o voo vai atrasar ou não) ou Regressão (tempo do atraso).*
- [✅] Comparar pelo menos dois algoritmos diferentes.
- [✅] Avaliar os modelos com métricas adequadas (Acurácia, F1-Score, RMSE, MAE, etc.).

### 🔍 Modelagem Não Supervisionada (Mínimo uma abordagem)
- [✅] Utilizar técnicas de **Clusterização** (ex: agrupar rotas, aeroportos ou companhias) ou **Redução de Dimensionalidade** (ex: PCA).
- [✅] Apresentar resultados visuais através de gráficos e fornecer a interpretação.

### 📝 Apresentação Crítica
- [✅] Compartilhar as principais conclusões do estudo.
- [✅] Discutir as limitações dos modelos desenvolvidos.
- [✅] Propor melhorias e próximos passos para o projeto.

---

## 5. Diferenciais (Vá além do básico)
Grupos que desejam se destacar podem explorar:
*   **Feature Engineering:** Criar variáveis como período do dia, feriados ou estações do ano.
*   **Análise Geográfica:** Criar mapas de rotas e atrasos por estado/aeroporto.
*   **Dashboards:** Construir painéis interativos para visualização.
*   **Técnicas Avançadas:** Detecção de anomalias ou aprendizado semi-supervisionado.

---

## 6. Perguntas Norteadoras
Para guiar sua análise, considere:
1. Quais aeroportos são mais críticos em relação a atrasos?
2. Que características aumentam a chance de atraso em um voo?
3. Os atrasos são mais comuns em certos dias da semana ou horários?
4. É possível agrupar aeroportos com perfis semelhantes?
5. Até que ponto conseguimos prever atrasos com base no histórico e contexto do voo?
