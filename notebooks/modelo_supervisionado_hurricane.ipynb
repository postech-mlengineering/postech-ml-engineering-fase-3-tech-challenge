```python
# Função haversine
import math

def haversine(lat1, lon1, lat2, lon2):
    R = 6371.0
    phi1, phi2 = math.radians(lat1), math.radians(lat2)
    dphi = math.radians(lat2 - lat1)
    dlambda = math.radians(lon2 - lon1)
    a = math.sin(dphi/2)**2 + math.cos(phi1)*math.cos(phi2)*math.sin(dlambda/2)**2
    return 2 * R * math.asin(math.sqrt(a))
```

```python
# Função para marcar voos próximos a furacões

def label_flight_with_hurricane(row, fur_df, days=1, radius_km=200):
    window_start = row['FLIGHT_DATE'] - pd.Timedelta(days=days)
    window_end   = row['FLIGHT_DATE'] + pd.Timedelta(days=days)
    mask_time = (fur_df['Date'] >= window_start) & (fur_df['Date'] <= window_end)
    fur_near = fur_df[mask_time]
    if fur_near.empty:
        return False
    dist_origin = haversine(row['ORIGIN_LATITUDE'], row['ORIGIN_LONGITUDE'],
                           fur_near['Latitude'], fur_near['Longitude'])
    dist_dest   = haversine(row['DEST_LATITUDE'], row['DEST_LONGITUDE'],
                           fur_near['Latitude'], fur_near['Longitude'])
    return (dist_origin.min() <= radius_km) or (dist_dest.min() <= radius_km)
```
