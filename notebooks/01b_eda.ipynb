```python
#bibliotecas
import pandas as pd
import plotly.graph_objects as go
from plotly.subplots import make_subplots

pd.set_option('display.max_columns', None)
```

```python
#dados
df_flights = pd.read_csv('../data/raw/flights.csv')
df_airlines = pd.read_csv('../data/raw/airlines.csv')
df_airports = pd.read_csv('../data/raw/airports.csv')
```

```python
#cruzamento dos dados
df = df_flights.merge(
    df_airlines.rename(columns={'AIRLINE': 'AIRLINE_NAME'}), 
    left_on='AIRLINE', right_on='IATA_CODE', how='left'
).drop(columns=['IATA_CODE'])

df_airports = df_airports.rename(columns={'AIRPORT': 'AIRPORT_NAME'})

df = df.merge(
    df_airports.add_prefix('ORIGIN_'), 
    left_on='ORIGIN_AIRPORT', right_on='ORIGIN_IATA_CODE', how='left'
).drop(columns=['ORIGIN_IATA_CODE'])

df = df.merge(
    df_airports.add_prefix('DESTINATION_'),
    left_on='DESTINATION_AIRPORT', right_on='DESTINATION_IATA_CODE', how='left'
).drop(columns=['DESTINATION_IATA_CODE'])

df.head()
```

**Obtendo target**

De acordo com o Departamento de Transportes dos EUA (BTS), um voo é considerado atrasado atrasado quando chega 15 ou mais
minutos depois do horário de chegada previsto. Diante disso, vamos adotar o padrão citado como target do modelo supervisionado.

Fonte: [https://www.transtats.bts.gov/ot_delay/ot_delaycause1.asp?20=E&2r4v1qs41z=FHGDE%20&2r4v1q61=FHGDG](https://www.transtats.bts.gov/ot_delay/ot_delaycause1.asp?20=E&2r4v1qs41z=FHGDE%20&2r4v1q61=FHGDG)

```python
#obtendo variável target
df['IS_DELAYED'] = (df['ARRIVAL_DELAY'] > 15).astype(int)
```

**Qual é a distribuição entre voos atrasados e pontuais?**

```python
df_temp = df['IS_DELAYED'].value_counts().reset_index()
df_temp.columns = ['TARGET_STATUS', 'COUNT']
df_temp['STATUS_NAME'] = df_temp['TARGET_STATUS'].map({0: 'Pontual', 1: 'Atrasado'})

colors_map = {'Pontual': '#3498db', 'Atrasado': '#e74c3c'}
colors = [colors_map[name] for name in df_temp['STATUS_NAME']]

fig = go.Figure(data=[go.Pie(
    labels=df_temp['STATUS_NAME'],
    values=df_temp['COUNT'],
    hole=0.4,
    pull=[0, 0.1],
    marker=dict(colors=colors),
    textinfo='percent+label+value'
)])
fig.update_layout(
    title='Distribuição dos Voos',
    template='plotly_white',
    height=600
)
fig.show()
```

**Qual o perfil dos atrasos?**

```python
delays = ['DEPARTURE_DELAY', 'ARRIVAL_DELAY']

df_temp = df[delays].describe()
df_temp.loc['cv'] = df_temp.loc['std'] / df_temp.loc['mean']
df_temp.loc['kurtosis'] = df[delays].kurtosis()
df_temp
```

```python
labels = ['Atraso de Partida', 'Atraso de Chegada']

fig = make_subplots(
    rows=3, cols=1,
    subplot_titles=(
        'Média de Minutos', 
        'Coeficiente de Variação', 
        'Curtose'
    ),
    vertical_spacing=0.1
)
fig.add_trace(go.Bar(
    x=labels, 
    y=df_temp.loc['mean', delays],
    marker=dict(color=df_temp.loc['mean', delays], colorscale='Blues'),
    text=df_temp.loc['mean', delays].round(2),
    textposition='auto',
    name='Média'
), row=1, col=1)
fig.add_trace(go.Bar(
    x=labels, 
    y=df_temp.loc['cv', delays],
    marker=dict(color=df_temp.loc['cv', delays], colorscale='Blues'),
    text=df_temp.loc['cv', delays].round(2),
    textposition='auto',
    name='CV'
), row=2, col=1)
fig.add_hline(
    y=1, line_dash='dash', line_color='red', row=2, col=1, 
    annotation_text='Alta Instabilidade', annotation_position='top left'
)
fig.add_trace(go.Bar(
    x=labels, 
    y=df_temp.loc['kurtosis', delays],
    marker=dict(color=df_temp.loc['kurtosis', delays], colorscale='Blues'),
    text=df_temp.loc['kurtosis', delays].round(0),
    textposition='outside',
    name='Curtose'
), row=3, col=1)
fig.update_layout(
    height=1000,
    title_text='Análise de Distribuição e Volatilidade dos Atrasos',
    template='plotly_white',
    showlegend=False
)
fig.update_yaxes(title_text='Tempo (min)', row=1, col=1)
fig.update_yaxes(title_text='CV', row=2, col=1)
fig.update_yaxes(title_text='Curtose', row=3, col=1)
fig.show()
```

Estatisticamente, a operação típica é pontual, com medianas negativas (-2 a -5 minutos) indicando que a maioria dos voos decola e pousa antes do previsto. No entanto, a média é positiva (9 minutos), sendo distorcida por atrasos catastróficos que chegam a 2.000 minutos. Essa discrepância confirma que a média aritmética não representa o comportamento real da maioria dos voos; por isso, a mediana será nossa referência para tratar dados faltantes, evitando que esses valores extremos "sujem" as informações típicas.

A instabilidade do sistema é comprovada por uma curtose altíssima (acima de 90) e um coeficiente de variação de 8.91. Na prática, isso significa que os atrasos não seguem uma distribuição "normal"; eles são eventos raros, mas quando ocorrem, são gigantescos (outliers extremos). 

Essa característica tem um impacto direto na escolha do nosso modelo: técnicas de padronização que dependem da média e do desvio padrão serão evitadas, pois seriam esmagadas por esses valores fora da curva. Em vez disso, priorizaremos modelos robustos baseados em árvores de decisão (como Random Forest e XGBoost), que são naturalmente resilientes a outliers e conseguem mapear padrões complexos de atraso sem serem enganados por casos extremos.

**Quais são as principais causas de atraso?**

```python
causes = ['AIR_SYSTEM_DELAY', 'AIRLINE_DELAY', 'LATE_AIRCRAFT_DELAY', 'SECURITY_DELAY', 'WEATHER_DELAY']

df_temp = df[causes].describe()
df_temp.loc['cv'] = df_temp.loc['std'] / df_temp.loc['mean']
df_temp.loc['kurtosis'] = df[causes].kurtosis()
df_temp
```

```python
labels = ['Sistema Aéreo', 'Companhia Aérea', 'Aeronave Tardia', 'Segurança', 'Clima']

fig = make_subplots(
    rows=3, cols=1,
    subplot_titles=(
        'Média de Minutos por Causa', 
        'Coeficiente de Variação por Causa', 
        'Curtose por Causa'
    ),
    vertical_spacing=0.1
)
fig.add_trace(go.Bar(
    x=labels, 
    y=df_temp.loc['mean', causes],
    marker=dict(color=df_temp.loc['mean', causes], colorscale='Blues'),
    text=df_temp.loc['mean', causes].round(2),
    textposition='auto',
    name='Média'
), row=1, col=1)
fig.add_trace(go.Bar(
    x=labels, 
    y=df_temp.loc['cv', causes],
    marker=dict(color=df_temp.loc['cv', causes], colorscale='Blues'),
    text=df_temp.loc['cv', causes].round(2),
    textposition='auto',
    name='CV'
), row=2, col=1)

fig.add_hline(
    y=1, line_dash='dash', line_color='red', row=2, col=1, 
    annotation_text='Alta Instabilidade', annotation_position='top left'
)
fig.add_trace(go.Bar(
    x=labels, 
    y=df_temp.loc['kurtosis', causes],
    marker=dict(color=df_temp.loc['kurtosis', causes], colorscale='Blues'),
    text=df_temp.loc['kurtosis', causes].round(0),
    textposition='outside',
    name='Curtose'
), row=3, col=1)
fig.update_layout(
    height=1000,
    title_text='Análise de Distribuição e Volatilidade das Causas de Atraso',
    template='plotly_white',
    showlegend=False
)
fig.update_yaxes(title_text='Tempo (min)', row=1, col=1)
fig.update_yaxes(title_text='CV', row=2, col=1)
fig.update_yaxes(title_text='Curtose', row=3, col=1)
fig.show()
```

A análise das causas revela que o maior desafio do setor é a eficiência operacional. O predomínio de atrasos por aeronave tardia evidencia um efeito cascata: o sistema não consegue absorver um atraso inicial, replicando-o nos voos seguintes. Como os atrasos por companhia aérea também são significativos, fica claro que a maior parte do tempo perdido decorre de falhas na gestão logística e manutenção, superando fatores externos inevitáveis como clima ou segurança.

O atraso por aeronave tardia apresenta a maior média (23,47 min), seguido de perto pelos atrasos de companhia aérea (18,96 min). Juntos, eles formam o núcleo de ineficiência operacional que o modelo precisa priorizar. Em contrapartida, fatores como Clima e Segurança apresentam médias baixas (2,91 e 0,07 min, respectivamente), mas curtoses extremamente elevadas — no caso da segurança, ultrapassando a marca de 10.000. Isso indica que, embora esses eventos externos sejam raros, quando ocorrem, geram impactos desproporcionais que fogem completamente à regra geral da operação, reforçando a necessidade de algoritmos que não sejam sensíveis apenas às médias, mas que compreendam a gravidade desses desvios extremos.

**Quando ocorrem os atrasos?**

```python
df_temp = df.copy()

df_temp['SCHEDULED_DEPARTURE_HOUR'] = df_temp['SCHEDULED_DEPARTURE'] // 100
df_temp['SCHEDULED_TIME_HOUR'] = df_temp['SCHEDULED_TIME_HOUR'] = df_temp['SCHEDULED_TIME'] / 60

df_month = df_temp.groupby('MONTH')['IS_DELAYED'].mean().reset_index()
df_dow = df_temp.groupby('DAY_OF_WEEK')['IS_DELAYED'].mean().reset_index()
df_hour = df_temp.groupby('SCHEDULED_DEPARTURE_HOUR')['IS_DELAYED'].mean().reset_index()

fig = make_subplots(
    rows=3, cols=1, 
    subplot_titles=('% Atraso por Mês', '% Atraso por Dia da Semana', '% Atraso por Hora do Dia'),
    vertical_spacing=0.1
)
fig.add_trace(
    go.Bar(x=df_month['MONTH'], y=df_month['IS_DELAYED'], marker_color='skyblue', name='Mês'),
    row=1, col=1
)
fig.add_trace(
    go.Bar(x=df_dow['DAY_OF_WEEK'], y=df_dow['IS_DELAYED'], marker_color='salmon', name='Dia Semana'),
    row=2, col=1
)
fig.add_trace(
    go.Scatter(x=df_hour['SCHEDULED_DEPARTURE_HOUR'], y=df_hour['IS_DELAYED'], mode='lines+markers', name='Hora'),
    row=3, col=1
)
fig.update_layout(
    height=900, 
    title_text='Análise Temporal de Atrasos',
    showlegend=False,
    template='plotly_white'
)
fig.update_xaxes(title_text='Mês', row=1, col=1)
fig.update_xaxes(title_text='Dia da Semana (1=Seg, 7=Dom)', row=2, col=1)
fig.update_xaxes(title_text='Hora de Partida Programada', row=3, col=1)
fig.show()
```

A análise temporal revela que a probabilidade de atraso não é uma variável estática, sendo fortemente regida por ciclos de demanda e pela degradação da eficiência operacional ao longo do dia. Observamos uma **sazonalidade bimodal** nos meses de fevereiro e junho, períodos marcados por condições climáticas extremas (neve e tempestades de verão) e picos de férias. No nível semanal, a pressão do tráfego corporativo é evidente, com os maiores índices de atraso concentrados às segundas e quintas-feiras, enquanto o sábado apresenta a maior estabilidade do sistema.

O diagnóstico mais crítico, contudo, surge na análise horária: a taxa de atraso escala progressivamente de menos de 10% às 05:00 para mais de 25% às 20:00. Este comportamento é a prova visual do **efeito cascata (knock-on effect)**: conforme o dia avança, o sistema perde sua capacidade de absorver desvios, transformando atrasos pontuais de voos matinais em um congestionamento sistêmico no período noturno. Esses achados validam a necessidade de incluir variáveis temporais no modelo e são corroborados pelas análises do Departamento de Transportes dos EUA (BTS).
