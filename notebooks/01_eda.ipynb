```python
#bibliotecas
import pandas as pd
import plotly.graph_objects as go
from plotly.subplots import make_subplots

pd.set_option('display.max_columns', None)
```

```python
#dados
df_flights = pd.read_csv('../data/raw/flights.csv')
df_airlines = pd.read_csv('../data/raw/airlines.csv')
df_airports = pd.read_csv('../data/raw/airports.csv')
```

```python
#cruzamento dos dados
df = df_flights.merge(
    df_airlines.rename(columns={'AIRLINE': 'AIRLINE_NAME'}), 
    left_on='AIRLINE', right_on='IATA_CODE', how='left'
).drop(columns=['IATA_CODE'])

df_airports = df_airports.rename(columns={'AIRPORT': 'AIRPORT_NAME'})

df = df.merge(
    df_airports.add_prefix('ORIGIN_'), 
    left_on='ORIGIN_AIRPORT', right_on='ORIGIN_IATA_CODE', how='left'
).drop(columns=['ORIGIN_IATA_CODE'])

df = df.merge(
    df_airports.add_prefix('DESTINATION_'),
    left_on='DESTINATION_AIRPORT', right_on='DESTINATION_IATA_CODE', how='left'
).drop(columns=['DESTINATION_IATA_CODE'])

df.head()
```

**Obtendo target**

De acordo com o Departamento de Transportes dos EUA (BTS), um voo é considerado atrasado atrasado quando chega 15 ou mais
minutos depois do horário de chegada previsto. Diante disso, vamos adotar o padrão citado como target do modelo supervisionado.

Fonte: [https://www.transtats.bts.gov/ot_delay/ot_delaycause1.asp?20=E&2r4v1qs41z=FHGDE%20&2r4v1q61=FHGDG](https://www.transtats.bts.gov/ot_delay/ot_delaycause1.asp?20=E&2r4v1qs41z=FHGDE%20&2r4v1q61=FHGDG)

```python
#obtendo variável target
df['IS_DELAYED'] = (df['ARRIVAL_DELAY'] > 15).astype(int)
```

**Qual é a distribuição entre voos atrasados e pontuais?**

```python
df_temp = df['IS_DELAYED'].value_counts().reset_index()
df_temp.columns = ['TARGET_STATUS', 'COUNT']
df_temp['STATUS_NAME'] = df_temp['TARGET_STATUS'].map({0: 'Pontual', 1: 'Atrasado'})

colors_map = {'Pontual': '#3498db', 'Atrasado': '#e74c3c'}
colors = [colors_map[name] for name in df_temp['STATUS_NAME']]

fig = go.Figure(data=[go.Pie(
    labels=df_temp['STATUS_NAME'],
    values=df_temp['COUNT'],
    hole=0.4,
    pull=[0, 0.1],
    marker=dict(colors=colors),
    textinfo='percent+label+value'
)])
fig.update_layout(
    title='Distribuição dos Voos',
    template='plotly_white',
    height=600
)
fig.show()
```

Observa-se que a base de dados é desbalanceada, com 82,4% de voos pontuais e apenas 17,6% de atrasados. Para evitar que o modelo fique viciado na classe majoritária e tenha dificuldade em prever os atrasos, aplicaremos under-sampling, selecionando uma amostra aleatória dos voos pontuais para igualar as proporções, garantindo que o algoritmo aprenda a identificar com a mesma eficiência tanto os voos no horário quanto aqueles que sofrem atrasos.

**Qual o perfil dos atrasos?**

Para realizar o diagnóstico de distribuição e volatilidade dos atrasos, utilizamos duas métricas estatísticas que permitem ir além da análise de resumos simples.

**1. Coeficiente de Variação (CV)**

O CV é uma medida de instabilidade relativa, calculada pela razão entre o desvio padrão ($\sigma$) e a média ($\mu$):

$$CV = \frac{\sigma}{\mu}$$

Diferente do desvio padrão isolado, o CV nos permite comparar a variabilidade de diferentes escalas. Ele é essencial para identificar a sobre-dispersão dos dados. No nosso gráfico, a linha vermelha representa o Limiar de Variabilidade Unitária (CV = 1.0). Valores acima deste limite indicam que a incerteza é maior que a própria média, revelando um sistema hiper-variável e imprevisível.

**2. Curtose (Kurtosis)**

A Curtose mede a concentração de massa nos extremos da distribuição, identificando o peso das "caudas". Sua fórmula utiliza a quarta potência da distância dos dados em relação à média:

$$K = \frac{\sum_{i=1}^{n} (x_i - \bar{x})^4 / n}{s^4}$$

Enquanto a média descreve o centro dos dados, a Curtose descreve os outliers. Ao elevar os desvios à quarta potência, esta métrica "castiga" matematicamente os valores extremos, permitindo-nos quantificar a intensidade dos chamados "Cisnes Negros" (atrasos raros, mas de magnitude catastrófica). 

**Justificativa**

A escolha destas métricas antes da análise de causas específicas é estratégica: 

1.  **Confiabilidade:** O CV nos avisa se a média é um valor confiável para previsões simples (no nosso caso, a alta instabilidade prova que não).
2.  **Direcionamento:** A curtose elevada (leptocúrtica) justifica tecnicamente o abandono de modelos lineares e a adoção de algoritmos de **Machine Learning robustos**, como as árvores de decisão e o XGBoost, que são treinados para processar essas anomalias sem sofrer o viés de distorção dos valores extremos. 

```python
delays = ['DEPARTURE_DELAY', 'ARRIVAL_DELAY']

df_temp = df[delays].describe()
df_temp.loc['cv'] = df_temp.loc['std'] / df_temp.loc['mean']
df_temp.loc['kurtosis'] = df[delays].kurtosis()
df_temp
```

```python
labels = ['Atraso de Partida', 'Atraso de Chegada']

fig = make_subplots(
    rows=3, cols=1,
    subplot_titles=(
        'Média de Minutos', 
        'Coeficiente de Variação', 
        'Curtose'
    ),
    vertical_spacing=0.1
)
fig.add_trace(go.Bar(
    x=labels, 
    y=df_temp.loc['mean', delays],
    marker=dict(color=df_temp.loc['mean', delays], colorscale='Blues'),
    text=df_temp.loc['mean', delays].round(2),
    textposition='auto',
    name='Média'
), row=1, col=1)
fig.add_trace(go.Bar(
    x=labels, 
    y=df_temp.loc['cv', delays],
    marker=dict(color=df_temp.loc['cv', delays], colorscale='Blues'),
    text=df_temp.loc['cv', delays].round(2),
    textposition='auto',
    name='CV'
), row=2, col=1)
fig.add_hline(
    y=1, line_dash='dash', line_color='red', row=2, col=1, 
    annotation_text='Limiar de Variabilidade Unitária', annotation_position='top left'
)
fig.add_trace(go.Bar(
    x=labels, 
    y=df_temp.loc['kurtosis', delays],
    marker=dict(color=df_temp.loc['kurtosis', delays], colorscale='Blues'),
    text=df_temp.loc['kurtosis', delays].round(0),
    textposition='outside',
    name='Curtose'
), row=3, col=1)
fig.update_layout(
    height=1000,
    title_text='Análise de Distribuição e Volatilidade dos Atrasos',
    template='plotly_white',
    showlegend=False
)
fig.update_yaxes(title_text='Tempo (min)', row=1, col=1)
fig.update_yaxes(title_text='CV', row=2, col=1)
fig.update_yaxes(title_text='Curtose', row=3, col=1)
fig.show()
```

Como evidenciado pelos dados, a operação típica é majoritariamente pontual, com medianas negativas (-2 a -5 minutos). No entanto, a média aritmética permanece positiva (~9 minutos na partida), sendo severamente distorcida por atrasos catastróficos que atingem a marca de 2.000 minutos. Essa discrepância confirma que a média falha em representar o comportamento central do sistema, o que justifica o uso da mediana como referência para o tratamento de dados faltantes, preservando a integridade das informações típicas contra o ruído dos valores extremos.

A instabilidade estrutural desse cenário é tecnicamente validada pelos indicadores de variabilidade e forma. O coeficiente de variação de 8.91 supera drasticamente o limiar de variabilidade unitária, caracterizando um estado de sobre-dispersão extrema, onde a incerteza domina a média. Complementarmente, a curtose superior a 90 confirma uma distribuição de cauda pesada, evidenciando a presença de outliers gigantescos. Na prática, esses números provam que os atrasos não seguem uma distribuição "normal", mas sim um padrão de eventos raros com impactos desproporcionais sobre a malha aérea.

Esses diagnósticos direcionam diretamente a arquitetura do modelo. Técnicas de padronização tradicionais seriam ineficazes, pois os cálculos de distância seriam "esmagados" pela magnitude desses desvios. Em vez disso, priorizaremos algoritmos robustos baseados em árvores de decisão, como Random Forest e XGBoost. Estes modelos são naturalmente resilientes a anomalias e capazes de mapear a complexidade sistêmica revelada pela alta curtose e pela hiper-variabilidade do CV, sem serem enviesados pelos casos extremos da distribuição.

**Quais são as principais causas de atraso?**

```python
causes = ['AIR_SYSTEM_DELAY', 'AIRLINE_DELAY', 'LATE_AIRCRAFT_DELAY', 'SECURITY_DELAY', 'WEATHER_DELAY']

df_temp = df[causes].describe()
df_temp.loc['cv'] = df_temp.loc['std'] / df_temp.loc['mean']
df_temp.loc['kurtosis'] = df[causes].kurtosis()
df_temp
```

```python
labels = ['Sistema Aéreo', 'Companhia Aérea', 'Aeronave Tardia', 'Segurança', 'Clima']

fig = make_subplots(
    rows=3, cols=1,
    subplot_titles=(
        'Média de Minutos por Causa', 
        'Coeficiente de Variação por Causa', 
        'Curtose por Causa'
    ),
    vertical_spacing=0.1
)
fig.add_trace(go.Bar(
    x=labels, 
    y=df_temp.loc['mean', causes],
    marker=dict(color=df_temp.loc['mean', causes], colorscale='Blues'),
    text=df_temp.loc['mean', causes].round(2),
    textposition='auto',
    name='Média'
), row=1, col=1)
fig.add_trace(go.Bar(
    x=labels, 
    y=df_temp.loc['cv', causes],
    marker=dict(color=df_temp.loc['cv', causes], colorscale='Blues'),
    text=df_temp.loc['cv', causes].round(2),
    textposition='auto',
    name='CV'
), row=2, col=1)

fig.add_hline(
    y=1, line_dash='dash', line_color='red', row=2, col=1, 
    annotation_text='Limiar de Variabilidade Unitária', annotation_position='top left'
)
fig.add_trace(go.Bar(
    x=labels, 
    y=df_temp.loc['kurtosis', causes],
    marker=dict(color=df_temp.loc['kurtosis', causes], colorscale='Blues'),
    text=df_temp.loc['kurtosis', causes].round(0),
    textposition='outside',
    name='Curtose'
), row=3, col=1)
fig.update_layout(
    height=1000,
    title_text='Análise de Distribuição e Volatilidade das Causas de Atraso',
    template='plotly_white',
    showlegend=False
)
fig.update_yaxes(title_text='Tempo (min)', row=1, col=1)
fig.update_yaxes(title_text='CV', row=2, col=1)
fig.update_yaxes(title_text='Curtose', row=3, col=1)
fig.show()
```

A análise das causas específicas revela que o maior desafio do setor reside na eficiência operacional interna. O predomínio de atrasos por aeronave tardia apresenta a maior média (23,47 min), seguido de perto pelos atrasos de companhia aérea (18,96 min). Juntos, esses fatores formam o núcleo de ineficiência logística da malha aérea. O protagonismo da aeronave tardia é a evidência estatística do efeito cascata: a incapacidade do sistema de absorver desvios iniciais, replicando o atraso de uma perna de voo para todas as decolagens subsequentes da mesma aeronave. 

Em contrapartida, fatores externos como clima e segurança apresentam médias nominais baixas (2,91 e 0,08 min, respectivamente), sugerindo um impacto rotineiro menor. Entretanto, a aplicação da métrica de Curtose nestas variáveis revela uma realidade distinta: os valores de curtose são extremamente elevados, ultrapassando a marca de 10.000 no caso da segurança. Isso caracteriza estes eventos como ocorrências raras que operam fora de qualquer padrão de normalidade, gerando impactos desproporcionais e catastróficos quando se manifestam. 

Essa dualidade reforça a estratégia de modelagem adotada. Enquanto o modelo precisa aprender a "tendência" dos atrasos logísticos (companhia aérea e aeronave tardia), ele também deve ser resiliente o suficiente para não ser desestabilizado pela hiper-variabilidade e pelos desvios extremos dos fatores externos. A escolha de algoritmos que compreendem a gravidade dos outliers, em vez de apenas seguir a média, é o que permitirá ao modelo separar o ruído operacional dos eventos sistêmicos críticos.

**Quando ocorrem os atrasos?**

```python
df_temp = df.copy()

df_temp['SCHEDULED_DEPARTURE_HOUR'] = df_temp['SCHEDULED_DEPARTURE'] // 100
df_temp['SCHEDULED_TIME_HOUR'] = df_temp['SCHEDULED_TIME_HOUR'] = df_temp['SCHEDULED_TIME'] / 60

df_month = df_temp.groupby('MONTH')['IS_DELAYED'].mean().reset_index()
df_dow = df_temp.groupby('DAY_OF_WEEK')['IS_DELAYED'].mean().reset_index()
df_hour = df_temp.groupby('SCHEDULED_DEPARTURE_HOUR')['IS_DELAYED'].mean().reset_index()

fig = make_subplots(
    rows=3, cols=1, 
    subplot_titles=('% Atraso por Mês', '% Atraso por Dia da Semana', '% Atraso por Hora do Dia'),
    vertical_spacing=0.1
)
fig.add_trace(
    go.Bar(x=df_month['MONTH'], y=df_month['IS_DELAYED'], marker_color='skyblue', name='Mês'),
    row=1, col=1
)
fig.add_trace(
    go.Bar(x=df_dow['DAY_OF_WEEK'], y=df_dow['IS_DELAYED'], marker_color='salmon', name='Dia Semana'),
    row=2, col=1
)
fig.add_trace(
    go.Scatter(x=df_hour['SCHEDULED_DEPARTURE_HOUR'], y=df_hour['IS_DELAYED'], mode='lines+markers', name='Hora'),
    row=3, col=1
)
fig.update_layout(
    height=900, 
    title_text='Análise Temporal de Atrasos',
    showlegend=False,
    template='plotly_white'
)
fig.update_xaxes(title_text='Mês', row=1, col=1)
fig.update_xaxes(title_text='Dia da Semana (1=Seg, 7=Dom)', row=2, col=1)
fig.update_xaxes(title_text='Hora de Partida Programada', row=3, col=1)
fig.show()
```

A análise temporal revela que a probabilidade de atraso não é uma variável estática, sendo fortemente regida por ciclos de demanda e pela degradação da eficiência operacional ao longo do dia. Observamos uma sazonalidade nos meses de fevereiro e junho, períodos marcados por condições climáticas extremas (neve e tempestades de verão) e picos de férias. No nível semanal, a pressão do tráfego corporativo é evidente, com os maiores índices de atraso concentrados às segundas e quintas-feiras, enquanto o sábado apresenta a maior estabilidade do sistema.

O diagnóstico mais crítico, contudo, surge na análise horária: a taxa de atraso escala progressivamente de menos de 10% às 05:00 para mais de 25% às 20:00. Este comportamento é a prova visual do efeito cascata: conforme o dia avança, o sistema perde sua capacidade de absorver desvios, transformando atrasos pontuais de voos matinais em um congestionamento sistêmico no período noturno. Esses achados validam a necessidade de incluir variáveis temporais no modelo e são corroborados pelas análises do Departamento de Transportes dos EUA (BTS).
